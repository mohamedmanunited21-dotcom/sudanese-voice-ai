## 5. محرك التفريغ اللغوي العميق والسردية (Whisper Large-v3) مع تحديد المشاعر والأنماط

```python
import whisper
import json

print('⏳ جاري تحميل نموذج Whisper Large-v3 للتعلم اللغوي العميق...')
model = whisper.load_model('large-v3', device='cuda' if torch.cuda.is_available() else 'cpu')
print('✅ Whisper Large-v3 جاهز للعمل بكامل طاقته!')

def transcribe_and_learn(audio_path, output_json_path):
    print(f'🧠 جاري الاستماع الكامل للعمل: {os.path.basename(audio_path)} ...')
    result = model.transcribe(
        audio_path,
        language='ar',
        task='transcribe',
        verbose=False,
        temperature=0.0
    )
    
    corpus_data = []
    for seg in result['segments']:
        corpus_data.append({
            'id': seg['id'],
            'start': seg['start'],
            'end': seg['end'],
            'text': seg['text'].strip(),
            'duration': seg['end'] - seg['start']
        })
    
    with open(output_json_path, 'w', encoding='utf-8') as f:
        json.dump(corpus_data, f, ensure_ascii=False, indent=2)
    
    print(f'✅ تم استخراج وتفريغ {len(corpus_data)} جملة حوارية وسردية كاملة.')
    return corpus_data
```

```python
import glob

# فحص واستيعاب جميع الأعمال الصوتية الـ 7 الكاملة تلقائياً
audio_files = glob.glob(os.path.join(BASE_DIR, 'corpus', '**', '*.wav'), recursive=True)
print(f'🎙️ تم العثور على {len(audio_files)} أعمال صوتية كاملة للبدء في الاستماع إليها:\n')

corpus_json_list = []
for audio in audio_files:
    fname = os.path.splitext(os.path.basename(audio))[0]
    out_json = os.path.join(BASE_DIR, 'corpus', f'{fname}_transcript.json')
    corpus_json_list.append(out_json)
    
    if os.path.exists(out_json):
        print(f'⏭️ هذا العمل مستوعب مسبقاً: {fname}')
        continue
        
    transcribe_and_learn(audio, out_json)

print('\n🎉 اكتمل الاستماع والتحليل اللغوي لجميع الأعمال بنجاح!')
```

## 6. استخراج قاموس المصطلحات السودانية والتنغيم الصوتي (Sudanese Lexicon & Prosody Learner)

```python
# كود استخراج معجم الكلمات السودانية المركبة والسياقية وتكرارها
from collections import Counter
import re

def extract_sudanese_lexicon(corpus_json_list):
    all_words = []
    for c_path in corpus_json_list:
        with open(c_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            for item in data:
                words = re.findall(r'\b[\u0600-\u06FF]+\b', item['text'])
                all_words.extend(words)
    
    counts = Counter(all_words)
    print(f'إجمالي الكلمات المعالجة في المعجم: {len(all_words)} كلمة')
    print('أكثر الكلمات السودانية تواتراً في السرديات:')
    for w, c in counts.most_common(25):
        print(f'{w}: {c}')

print('✅ محرك بناء المعجم اللغوي جاهز.')
```

```python
# تشغيل خط الإنتاج المنهجي الشامل (قناة تلو الأخرى)
# Whisper Large-v3 مفعل مسبقاً في المتغير model
# -*- coding: utf-8 -*-
"""
🇸🇩 نظام الفهرسة والتفريغ المنهجي لكل قناة سودانية (Systematic Channel-by-Channel Pipeline)
يقوم هذا النظام بمعالجة القنوات قناة تلو الأخرى:
1. فتح القناة وحصر 20 مقطعاً/فيديو كاملاً.
2. تحميل وتفريغ جميع مقاطع القناة بالكامل عبر Whisper Large-v3.
3. استخراج معجم ومصطلحات القناة وخصائصها الصوتية.
4. الانتقال التلقائي إلى القناة التالية بعد إتمام القناة الحالية بنسبة 100%.
"""

import os
import sys
import json
import subprocess
import glob
from collections import Counter
import re

# ضبط الترميز لدعم العربية في سطر الأوامر
if sys.platform == 'win32':
    try:
        sys.stdout.reconfigure(encoding='utf-8')
    except Exception:
        pass

# المسار الرئيسي على Google Drive
if os.path.exists('/content/drive/My Drive/Sudanese_Voice_Project'):
    BASE_DIR = '/content/drive/My Drive/Sudanese_Voice_Project'
else:
    BASE_DIR = r'G:\My Drive\Sudanese_Voice_Project'

CORPUS_DIR = os.path.join(BASE_DIR, 'corpus')
MODELS_DIR = os.path.join(BASE_DIR, 'models')
os.makedirs(CORPUS_DIR, exist_ok=True)
os.makedirs(MODELS_DIR, exist_ok=True)

# 1. قائمة القنوات المعتمدة بالترتيب المنهجي
CHANNELS = [
    {
        'id': 'azouzfhama',
        'name': 'عزوز فهامة (عمارقة الكوميديا والدراما الشعبية)',
        'category': '4_drama_and_chat',
        'url': 'https://www.youtube.com/@azouzfhama3670/videos',
        'target_count': 20
    },
    {
        'id': 'alsudanpodcast',
        'name': 'بودكاست السودان (نقاش فكري استوديو معمق)',
        'category': '3_intellectual_podcasts',
        'url': 'https://www.youtube.com/@alsudanpodcast/videos',
        'target_count': 20
    },
    {
        'id': 'lq_media',
        'name': 'لقاء ميديا (ونسة وفضفضة وحوار شبابي اجتماعي)',
        'category': '4_drama_and_chat',
        'url': 'https://www.youtube.com/@Lq_media/videos',
        'target_count': 20
    },
    {
        'id': 'salmasayed',
        'name': 'سلمى سيد (حوارات وثائقية ونبرة نسائية راقية)',
        'category': '2_female_interviews',
        'url': 'https://www.youtube.com/@salmasayed7401/videos',
        'target_count': 20
    },
    {
        'id': 'elfatiherwa',
        'name': 'الفاتح عروة (سرد تاريخي وسياسي ورواية هادئة)',
        'category': '3_intellectual_podcasts',
        'url': 'https://www.youtube.com/@elfatiherwa/videos',
        'target_count': 20
    },
    {
        'id': 'shailaban',
        'name': 'شيلابن (Shailaban - بودكاست وكوميديا وسخرية)',
        'category': '4_drama_and_chat',
        'url': 'https://www.youtube.com/@shailaban8325/videos',
        'target_count': 20
    },
    {
        'id': 'sudanesedrama',
        'name': 'الدراما والمسلسلات السودانية (بلة والكاشف ولغة الشارع والبيت)',
        'category': '4_drama_and_chat',
        'url': 'https://www.youtube.com/@SudaneseDrama/videos',
        'target_count': 20
    },
    {
        'id': 'profm',
        'name': 'إذاعة PRO 106.6 FM (بث إذاعي حي وتفاعل سريع)',
        'category': '1_radio_tv_formal',
        'url': 'https://www.youtube.com/@PRO106.6fm/videos',
        'target_count': 20
    },
    {
        'id': 'sudania24',
        'name': 'تلفزيون سودانية 24 (حوار تلفزيوني رصين ومخارج واضحة)',
        'category': '1_radio_tv_formal',
        'url': 'https://www.youtube.com/@Sudania24TV/videos',
        'target_count': 20
    },
    {
        'id': 'abubakr_faisal',
        'name': 'أبوبكر فيصل (ونسة وفضفضة سودانية بحتة)',
        'category': '4_drama_and_chat',
        'url': 'https://www.youtube.com/@AbubakrFaisal25/videos',
        'target_count': 20
    }
]

def fetch_channel_video_list(channel_url, max_count=20):
    """جلب قائمة المقاطع الكاملة من القناة"""
    print(f"🔍 فحص القناة واستخراج قائمة أفضل {max_count} مقطعاً...")
    cmd = [
        'yt-dlp',
        '--flat-playlist',
        '-j',
        '--playlist-items', f'1-{max_count}',
        channel_url
    ]
    try:
        proc = subprocess.run(cmd, capture_output=True, text=True, check=True)
        videos = []
        for line in proc.stdout.strip().split('\n'):
            if not line:
                continue
            try:
                item = json.loads(line)
                dur_min = round(item.get('duration', 0) / 60)
                videos.append({
                    'id': item.get('id'),
                    'title': item.get('title', 'فيديو'),
                    'duration_min': dur_min
                })
            except Exception:
                continue
        return videos
    except Exception as e:
        print(f"❌ خطأ أثناء فحص القناة: {e}")
        return []

def download_audio_file(video_id, save_path):
    """تحميل واستخلاص الصوت بجودة 24kHz Mono"""
    if os.path.exists(save_path):
        return True
    
    url = f"https://www.youtube.com/watch?v={video_id}"
    cmd = [
        'yt-dlp',
        '-x',
        '--audio-format', 'wav',
        '--audio-quality', '0',
        '--postprocessor-args', '-ar 24000 -ac 1',
        '-o', save_path,
        url
    ]
    try:
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        return os.path.exists(save_path)
    except Exception as e:
        print(f"  ❌ تعذر تحميل المقطع {video_id}: {e}")
        return False

def transcribe_with_whisper(model, audio_path, transcript_path):
    """تفريغ صوتي ولغوي كامل عبر Whisper Large-v3"""
    if os.path.exists(transcript_path):
        return True
    
    print(f"  🧠 الاستماع والتحليل اللساني للملف...")
    try:
        result = model.transcribe(
            audio_path,
            language='ar',
            task='transcribe',
            verbose=False,
            temperature=0.0
        )
        
        segments = []
        for s in result.get('segments', []):
            segments.append({
                'id': s['id'],
                'start': s['start'],
                'end': s['end'],
                'text': s['text'].strip(),
                'duration': s['end'] - s['start']
            })
            
        with open(transcript_path, 'w', encoding='utf-8') as f:
            json.dump(segments, f, ensure_ascii=False, indent=2)
            
        print(f"  ✅ تم بنجاح تفريغ {len(segments)} جملة وحوار صوتي.")
        return True
    except Exception as e:
        print(f"  ❌ خطأ أثناء التفريغ: {e}")
        return False

def run_systematic_pipeline(whisper_model=None):
    """تشغيل الفهرسة والتفريغ المنهجي قناة تلو الأخرى"""
    print("\n" + "="*70)
    print("🇸🇩 بدء نظام الفهرسة والتفريغ المنهجي للقنوات السودانية (20 مقطعاً لكل قناة)")
    print("="*70)

    for ch_idx, ch in enumerate(CHANNELS, 1):
        print(f"\n========================================================")
        print(f"📺 [القناة {ch_idx}/{len(CHANNELS)}]: {ch['name']}")
        print(f"🔗 الرابط: {ch['url']}")
        print(f"🎯 المستهدف: {ch['target_count']} مقطعاً كاملاً")
        print(f"========================================================")

        ch_dir = os.path.join(CORPUS_DIR, ch['category'])
        os.makedirs(ch_dir, exist_ok=True)

        videos = fetch_channel_video_list(ch['url'], ch['target_count'])
        print(f"📋 تم حصر {len(videos)} مقطعاً من القناة للبدء في معالجتها بالتتابع:\n")

        channel_words = []
        completed_in_ch = 0

        for v_idx, v in enumerate(videos, 1):
            clean_title = re.sub(r'[\\/*?:"<>|]', '', v['title'])[:40].strip()
            base_fname = f"{ch['id']}_{v_idx:02d}_{v['id']}"
            audio_file = os.path.join(ch_dir, f"{base_fname}.wav")
            transcript_file = os.path.join(ch_dir, f"{base_fname}_transcript.json")

            print(f"[{v_idx}/{len(videos)}] 🎬 {clean_title} ({v['duration_min']} دقيقة)")

            # 1. التنزيل
            if not os.path.exists(audio_file):
                print(f"  ⬇️ جاري سحب الصوت وحفظه في Google Drive...")
                success = download_audio_file(v['id'], audio_file)
                if not success:
                    continue
            else:
                print(f"  ⏭️ الصوت موجود ومحفوظ مسبقاً في Drive.")

            # 2. التفريغ
            if whisper_model is not None:
                if not os.path.exists(transcript_file):
                    transcribe_with_whisper(whisper_model, audio_file, transcript_file)
                else:
                    print(f"  ⏭️ التفريغ مكتمل مسبقاً.")

            # 3. قراءة المعجم
            if os.path.exists(transcript_file):
                try:
                    with open(transcript_file, 'r', encoding='utf-8') as f:
                        data = json.load(f)
                        for item in data:
                            words = re.findall(r'\b[\u0600-\u06FF]+\b', item.get('text', ''))
                            channel_words.extend(words)
                    completed_in_ch += 1
                except Exception:
                    pass

        print(f"\n🏆 اكتملت معالجة القناة بالكامل: {ch['name']} ({completed_in_ch}/{len(videos)} مقطعاً)")
        if channel_words:
            ch_counts = Counter(channel_words)
            print(f"📚 إجمالي كلمات القناة: {len(channel_words)} كلمة | مفردات مميزة: {len(ch_counts)}")
            print("🔹 أكثر 10 مفردات تواتراً في هذه القناة:")
            for w, c in ch_counts.most_common(10):
                print(f"   - {w}: {c}")

        print(f"\n➡️ الانتقال الآن تلقائياً إلى القناة التالية...\n")

    print("\n🎉🎉 أُنجزت جميع القنوات العشر بنجاح تام وتم تحديث Google Drive بالكامل! 🎉🎉")

run_systematic_pipeline(whisper_model=model)
```
