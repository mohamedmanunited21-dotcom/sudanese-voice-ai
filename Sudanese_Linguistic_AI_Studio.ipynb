## 4. تحميل الحلقات والأعمال الكاملة بأعلى جودة صوتية مباشرة إلى Google Drive

```python
import subprocess

def download_full_episode(video_url, category, custom_name=None):
    target_dir = GENRES.get(category, GENRES['drama_and_chat'])
    output_template = os.path.join(target_dir, '%(title)s_%(id)s.%(ext)s' if not custom_name else f'{custom_name}.%(ext)s')
    
    cmd = [
        'yt-dlp',
        '-x',
        '--audio-format', 'wav',
        '--audio-quality', '0',
        '--postprocessor-args', '-ar 24000 -ac 1',
        '-o', output_template,
        video_url
    ]
    print(f'⏳ جاري سحب العمل الكامل من: {video_url} ...')
    subprocess.run(cmd, check=True)
    print(f'✅ تم الحفظ بنجاح في: {target_dir}')

# أمثلة للحلقات والأعمال المحددة من قبلك:
# 1. حلقة سلمى سيد كاملة (نبرة نسائية راقية)
# download_full_episode('https://www.youtube.com/watch?v=sUuuYDsDuTg', 'female_interviews', 'salma_sayed_special')
# 2. عمل كامل من Lq Media
# download_full_episode('https://www.youtube.com/watch?v=Mpt6yH55vJQ', 'drama_and_chat', 'lq_media_full_episode')
# 3. لقاء إذاعي كامل PRO FM
# download_full_episode('https://www.youtube.com/watch?v=b4B_UrKoTUA', 'radio_tv_formal', 'pro_fm_broadcast_full')
```

## 5. محرك التفريغ اللغوي العميق والسردية (Whisper Large-v3) مع تحديد المشاعر والأنماط

```python
import whisper
import json

print('⏳ جاري تحميل نموذج Whisper Large-v3 للتعلم اللغوي العميق...')
model = whisper.load_model('large-v3', device='cuda' if torch.cuda.is_available() else 'cpu')
print('✅ Whisper Large-v3 جاهز للعمل بكامل طاقته!')

def transcribe_and_learn(audio_path, output_json_path):
    print(f'🧠 جاري الاستماع الكامل للعمل: {os.path.basename(audio_path)} ...')
    result = model.transcribe(
        audio_path,
        language='ar',
        task='transcribe',
        verbose=False,
        temperature=0.0
    )
    
    corpus_data = []
    for seg in result['segments']:
        corpus_data.append({
            'id': seg['id'],
            'start': seg['start'],
            'end': seg['end'],
            'text': seg['text'].strip(),
            'duration': seg['end'] - seg['start']
        })
    
    with open(output_json_path, 'w', encoding='utf-8') as f:
        json.dump(corpus_data, f, ensure_ascii=False, indent=2)
    
    print(f'✅ تم استخراج وتفريغ {len(corpus_data)} جملة حوارية وسردية كاملة.')
    return corpus_data
```

## 6. استخراج قاموس المصطلحات السودانية والتنغيم الصوتي (Sudanese Lexicon & Prosody Learner)

```python
# كود استخراج معجم الكلمات السودانية المركبة والسياقية وتكرارها
from collections import Counter
import re

def extract_sudanese_lexicon(corpus_json_list):
    all_words = []
    for c_path in corpus_json_list:
        with open(c_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            for item in data:
                words = re.findall(r'\b[\u0600-\u06FF]+\b', item['text'])
                all_words.extend(words)
    
    counts = Counter(all_words)
    print(f'إجمالي الكلمات المعالجة في المعجم: {len(all_words)} كلمة')
    print('أكثر الكلمات السودانية تواتراً في السرديات:')
    for w, c in counts.most_common(25):
        print(f'{w}: {c}')

print('✅ محرك بناء المعجم اللغوي جاهز.')
```
